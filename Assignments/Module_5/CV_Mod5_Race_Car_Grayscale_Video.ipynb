```python
import cv2
from moviepy.editor import VideoFileClip

input_video = 'race_car.mp4'
output_video = 'race_car_gray.mp4'

# Step 1: Create VideoCapture object
cap = cv2.VideoCapture(input_video)

# Step 2: Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Step 3: Create VideoWriter
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video, fourcc, fps, (width, height))

frame_count = 0

# Step 4: Processing loop
while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    frame_count += 1

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Bonus: improve contrast
    gray = cv2.equalizeHist(gray)

    # Convert back to BGR (VERY IMPORTANT)
    gray_bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

    # Add text banner
    text = f'Grayscale - Frame: {frame_count}'
    cv2.putText(gray_bgr, text, (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 1,
                (255, 255, 255), 2)

    # Write frame
    out.write(gray_bgr)

# Step 5: Release
cap.release()
out.release()
cv2.destroyAllWindows()

# Step 6: Preview
clip = VideoFileClip(output_video)
clip.ipython_display(width=600)
```
