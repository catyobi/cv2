```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img_circle = cv2.imread('cv2/Mod3/shapes.jpg')

plt.figure(figsize=(6,6))
plt.imshow(img_circle[:,:,::-1])
plt.title("Original Image")
plt.axis("off")
```

```python
print(img_circle.shape)
```

```python
img_gray_circle = cv2.cvtColor(img_circle, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6,6))
plt.imshow(img_gray_circle, cmap='gray')
plt.title("Gray Scale Image")
plt.axis("off")
```

```python
print(img_gray_circle.shape)
```

```python
img_gray_circle = cv2.cvtColor(img_circle, cv2.COLOR_BGR2GRAY)
```

```python
plt.figure(figsize=(6,6))
plt.imshow(img_gray_circle, cmap='gray')
plt.title("Gray Scale Image")
plt.axis("off")
```

```python
retval, img_ring_mask = cv2.threshold(img_gray_circle, 200, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(6,6))
plt.imshow(img_ring_mask, cmap='gray')
plt.title("Mask")
plt.axis("off")
```

```python
img_ring_mask_inv = cv2.bitwise_not(img_ring_mask)
plt.figure(figsize=(6,6))
plt.imshow(img_ring_mask_inv, cmap='gray')
plt.title("Inverse Mask")
plt.axis("off")
```

```python
retval, img_ring_mask = cv2.threshold(img_gray_circle, 230, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(6,6))
plt.imshow(img_ring_mask, cmap='gray')
plt.title("Mask (threshold=230)")
plt.axis("off")
```

```python
img_combined = cv2.bitwise_and(img_circle, img_circle, mask=img_ring_mask_inv)

plt.figure(figsize=(6,6))
plt.imshow(img_combined[:,:,::-1])
plt.title("Combined Image")
plt.axis("off")
```

```python
plt.figure(figsize = [20,10])
plt.subplot(141); plt.imshow(img_gray_circle);          plt.title('Gray Scale Image')
plt.subplot(142); plt.imshow(img_ring_mask);            plt.title('Mask')
plt.subplot(143); plt.imshow(img_ring_mask_inv);        plt.title('Inverse Mask')
plt.subplot(144); plt.imshow(img_combined[:, :, ::-1]); plt.title('Combined')
```
